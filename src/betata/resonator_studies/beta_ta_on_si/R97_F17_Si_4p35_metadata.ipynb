### Add fabrication, simulation, and measurement metadata for R97_F17_Si_4p35

```python
from pathlib import Path
import pandas as pd
from betata.resonator_studies.resonator import Resonator, save_resonator, add_inductance_metadata
CWD = Path.cwd()
```

```python
SPR_SIM_FILEPATH = CWD / "data/beta_ta_on_si/spr_sim_si.csv"

def add_spr_metadata(resonator: Resonator):
    """ """
    df = pd.read_csv(SPR_SIM_FILEPATH)
    pitch_um = round(resonator.pitch * 1e6)
    metadata_row = df.loc[df["pitch (um)"] == pitch_um]
    (resonator.width,) = metadata_row["width (um)"] * 1e-6
    (resonator.p_ms,) = metadata_row["p_ms"]
    (resonator.p_ma,) = metadata_row["p_ma"]
    (resonator.p_sa,) = metadata_row["p_sa"]
    (resonator.p_sub,) = metadata_row["p_sub"]
    return resonator
```

```python
resonator_name = "R97_F17_Si_4p35"
resonator_file = CWD / f"out/beta_ta_on_si/{resonator_name}.h5"
```

```python
resonator = Resonator(
    name=resonator_name,
    type="CPW",
    design_name="TAHP02KI-200",
    cooldown_name="2025-12-22_BFD",
    film_thickness=200e-9,
    pitch=2e-6,
    length=2100.16e-6,
    line_attenuation=89.68,
)
resonator = add_spr_metadata(resonator)
resonator = add_inductance_metadata(resonator)
```

```python
save_resonator(resonator, resonator_file)
```
